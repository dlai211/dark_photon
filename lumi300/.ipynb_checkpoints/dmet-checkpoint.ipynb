```python
def getVarDict(fb, process, var_name=None):
    var_dict = {}
    if var_name is None or var_name == 'dmet':
        var_dict['dmet'] = {
            'var': fb['dmet'],
            'bins': np.linspace(-100000, 100000, 20+1),
            'title': r'$E_{T,\mathrm{noJVT}}^{miss}-E_T^{miss}\ [\mathrm{MeV}]$',
            'shift': '*1'
        }
    return var_dict


def scale_good_bad_weights(w_good, w_bad, pct, mode="improvement"):
    """
    Scales good and bad vertex weights while preserving total process yield.
    - mode="improvement": pct=0 is baseline; pct=100 converts 100% of bad PV into good PV.
    - mode="exact_fraction": pct=0 is 100% bad PV; pct=100 is 100% good PV.
    """
    f = pct / 100.0
    sum_g = np.sum(w_good)
    sum_b = np.sum(w_bad)
    tot_w = sum_g + sum_b

    if tot_w == 0:
        return w_good, w_bad

    if mode == "improvement":
        # Convert fraction f of bad vertices into good vertices
        target_b = sum_b * (1.0 - f)
        target_g = tot_w - target_b
    elif mode == "exact_fraction":
        # Force exact fraction f of total yield to be good vertices
        target_g = tot_w * f
        target_b = tot_w * (1.0 - f)

    scale_g = (target_g / sum_g) if sum_g > 0 else 0.0
    scale_b = (target_b / sum_b) if sum_b > 0 else 0.0

    return w_good * scale_g, w_bad * scale_b


def run_dmet_study(
    tot_array,
    apply_vertex_bdt=False,
    scaling_mode="improvement",
    pv_pcts=range(0, 101, 10),
    dmet_low_cuts=[-100000],   # Add your website slider grid values here!
    dmet_high_cuts=[100000]    # Add your website slider grid values here!
):
    # 1. Select output folder based on whether VertexBDTScore cut is applied
    folder_tag = "with_vertexbdt" if apply_vertex_bdt else "without_vertexbdt"
    base_out_dir = f"/home/jlai/dark_photon/lumi300/plots_{folder_tag}"
    os.makedirs(base_out_dir, exist_ok=True)

    # 2. Extract Signal (index 0: ggHyyd) and Background (index 1: gammajet_direct)
    fb_sig = tot_array[0]
    fb_bkg = tot_array[1]

    if apply_vertex_bdt:
        fb_sig = fb_sig[fb_sig['VertexBDTScore'] > 0.1]
        fb_bkg = fb_bkg[fb_bkg['VertexBDTScore'] > 0.1]

    var = 'dmet'
    var_config = getVarDict(fb_sig, 'ggHyyd', var_name=var)
    bins = var_config[var]['bins']
    title = var_config[var]['title']

    # Convert to flat numpy arrays once for fast slider loops
    sig_x = ak.to_numpy(fb_sig[var])
    sig_w = ak.to_numpy(fb_sig['weights'])
    sig_good = ak.to_numpy(fb_sig['is_good_pv'])

    bkg_x = ak.to_numpy(fb_bkg[var])
    bkg_w = ak.to_numpy(fb_bkg['weights'])
    bkg_good = ak.to_numpy(fb_bkg['is_good_pv'])

    # Separate into 4 components: (Sig Good, Sig Bad, Bkg Good, Bkg Bad)
    sig_x_good, sig_w_good_0 = sig_x[sig_good], sig_w[sig_good]
    sig_x_bad,  sig_w_bad_0  = sig_x[~sig_good], sig_w[~sig_good]

    bkg_x_good, bkg_w_good_0 = bkg_x[bkg_good], bkg_w[bkg_good]
    bkg_x_bad,  bkg_w_bad_0  = bkg_x[~bkg_good], bkg_w[~bkg_good]

    # Four distinct colors
    colors = {
        "bkg_good": "#4C72B0",  # Steel Blue (Background Good PV)
        "bkg_bad":  "#DD8452",  # Coral Orange (Background Bad PV)
        "sig_good": "#D62728",  # Crimson Red (Signal Good PV)
        "sig_bad":  "#9467BD"   # Amethyst Purple (Signal Bad PV)
    }

    # 3. Loop over Slider 1: True Vertex % (0% to 100% in steps of 10%)
    for pct in pv_pcts:
        sig_w_good, sig_w_bad = scale_good_bad_weights(sig_w_good_0, sig_w_bad_0, pct, mode=scaling_mode)
        bkg_w_good, bkg_w_bad = scale_good_bad_weights(bkg_w_good_0, bkg_w_bad_0, pct, mode=scaling_mode)

        # Combined arrays at preselection (before dmet window cut)
        sig_all_x = np.concatenate([sig_x_good, sig_x_bad])
        sig_all_w = np.concatenate([sig_w_good, sig_w_bad])
        bkg_all_x = np.concatenate([bkg_x_good, bkg_x_bad])
        bkg_all_w = np.concatenate([bkg_w_good, bkg_w_bad])

        # 4. Loop over Sliders 2 & 3: Lower and Upper cuts on dmet
        for d_low in dmet_low_cuts:
            for d_high in dmet_high_cuts:
                if d_low >= d_high:
                    continue

                # Masks for the [d_low, d_high] window
                m_sig_all = (sig_all_x >= d_low) & (sig_all_x <= d_high)
                m_bkg_all = (bkg_all_x >= d_low) & (bkg_all_x <= d_high)

                # --- A. Performance & Significance Plot ---
                fig, (ax_top, ax_bot) = plt.subplots(
                    2, 1, figsize=(12, 13), gridspec_kw={'height_ratios': [9, 4]}
                )

                # Top Plot: Stacked Background (Good PV + Bad PV)
                ax_top.hist(
                    [bkg_x_bad, bkg_x_good],
                    bins=bins,
                    weights=[bkg_w_bad, bkg_w_good],
                    color=[colors["bkg_bad"], colors["bkg_good"]],
                    label=[
                        rf'$\gamma$+jets (Bad PV) [$\Sigma w$: {np.sum(bkg_w_bad):.0f}]', 
                        rf'$\gamma$+jets (Good PV) [$\Sigma w$: {np.sum(bkg_w_good):.0f}]'
                    ],
                    stacked=True,
                    alpha=0.85
                )

                # Top Plot: Overlaid Step Histograms for Signal (Good PV & Bad PV)
                ax_top.hist(
                    sig_x_good, bins=bins, weights=sig_w_good,
                    color=colors["sig_good"], 
                    label=rf'ggHyyd (Good PV) [$\Sigma w$: {np.sum(sig_w_good):.1f}]',
                    histtype='step', linewidth=2.5
                )
                ax_top.hist(
                    sig_x_bad, bins=bins, weights=sig_w_bad,
                    color=colors["sig_bad"],
                    label=rf'ggHyyd (Bad PV) [$\Sigma w$: {np.sum(sig_w_bad):.1f}]',
                    histtype='step', linewidth=2.5, linestyle='--'
                )

                # Error bars on Good and Bad Signal
                for sx, sw, scol in [
                    (sig_x_good, sig_w_good, colors["sig_good"]),
                    (sig_x_bad, sig_w_bad, colors["sig_bad"])
                ]:
                    if len(sx) > 0 and np.sum(sw) > 0:
                        s_counts, bin_edges = np.histogram(sx, bins=bins, weights=sw)
                        sum_w_sq, _ = np.histogram(sx, bins=bins, weights=sw**2)
                        bin_centers = 0.5 * (bin_edges[:-1] + bin_edges[1:])
                        ax_top.errorbar(
                            bin_centers, s_counts, yerr=np.sqrt(sum_w_sq),
                            fmt='.', linewidth=1.5, color=scol, capsize=0
                        )

                # Draw vertical cut lines if slider cuts are inside the plot range
                if d_low > bins[0]:
                    ax_top.axvline(d_low, color='black', linestyle='-.', linewidth=2, label=f'Cut Low: {d_low/1000:.0f} GeV')
                    ax_bot.axvline(d_low, color='black', linestyle='-.', linewidth=2)
                if d_high < bins[-1]:
                    ax_top.axvline(d_high, color='black', linestyle=':', linewidth=2, label=f'Cut High: {d_high/1000:.0f} GeV')
                    ax_bot.axvline(d_high, color='black', linestyle=':', linewidth=2)

                ax_top.set_yscale('log')
                ax_top.set_ylim(1e-4, 1e11)
                ax_top.set_xlim(bins[0], bins[-1])
                ax_top.minorticks_on()
                ax_top.grid(True, which="both", linestyle="--", linewidth=0.5)
                ax_top.set_ylabel("Events")
                ax_top.set_title(
                    f"dmet ({folder_tag.replace('_', ' ')}) | Good PV Scale: {pct}% | "
                    f"Window: [{d_low/1000:.0f}, {d_high/1000:.0f}] GeV"
                )
                ax_top.legend(ncol=2)

                # --- B. Bottom Plot: Per-Bin & Windowed Total Significance ---
                S_counts, _ = np.histogram(sig_all_x, bins=bins, weights=sig_all_w)
                B_counts, _ = np.histogram(bkg_all_x, bins=bins, weights=bkg_all_w)

                sqrt_B = np.sqrt(B_counts)
                sqrt_SplusB = np.sqrt(S_counts + B_counts)
                sqrt_Splus1p3B = np.sqrt(S_counts + 1.3 * B_counts)

                sig_simple = np.where(B_counts > 0, S_counts / sqrt_B, 0)
                sig_s_plus_b = np.where((S_counts + B_counts) > 0, S_counts / sqrt_SplusB, 0)
                sig_s_plus_1p3b = np.where((S_counts + 1.3 * B_counts) > 0, S_counts / sqrt_Splus1p3B, 0)
                zbi_per_bin = np.array([
                    zbi(S_counts[i], B_counts[i], sigma_b_frac=0.3) for i in range(len(S_counts))
                ])

                bin_centers = 0.5 * (bins[:-1] + bins[1:])

                # Total significance calculated STRICTLY inside [d_low, d_high]
                total_signal = np.sum(sig_all_w[m_sig_all])
                total_bkg = np.sum(bkg_all_w[m_bkg_all])

                if total_bkg > 0 and total_signal > 0:
                    total_sig_simple = total_signal / np.sqrt(total_bkg)
                    total_sig_s_plus_b = total_signal / np.sqrt(total_signal + total_bkg)
                    total_sig_s_plus_1p3b = total_signal / np.sqrt(total_signal + 1.3 * total_bkg)
                    total_sig_binomial = zbi(total_signal, total_bkg, sigma_b_frac=0.3)
                else:
                    total_sig_simple = total_sig_s_plus_b = total_sig_s_plus_1p3b = total_sig_binomial = 0.0

                ax_bot.step(bin_centers, sig_simple, where='mid', color='chocolate', linewidth=2,
                            label=f"S/√B = {total_sig_simple:.4f}")
                ax_bot.step(bin_centers, sig_s_plus_b, where='mid', color='tomato', linewidth=2,
                            label=f"S/√(S+B) = {total_sig_s_plus_b:.4f}")
                ax_bot.step(bin_centers, sig_s_plus_1p3b, where='mid', color='orange', linewidth=2,
                            label=f"S/√(S+1.3B) = {total_sig_s_plus_1p3b:.4f}")
                ax_bot.step(bin_centers, zbi_per_bin, where='mid', color='plum', linewidth=2,
                            label=f"Binomial ExpZ = {total_sig_binomial:.4f}")

                ax_bot.set_xlabel(title)
                ax_bot.set_ylabel("Significance")
                ax_bot.set_xlim(bins[0], bins[-1])
                ax_bot.grid(True, which="both", linestyle="--", linewidth=0.5)

                leg = ax_bot.legend()
                for text in leg.get_texts():
                    text.set_color('purple')

                plt.tight_layout()
                file_tag = f"{var}_pv{pct:03d}_low{int(d_low/1000)}_high{int(d_high/1000)}"
                path1 = f"{base_out_dir}/{file_tag}.png"
                plt.savefig(path1, dpi=150)
                plt.close()

                # --- C. ROC Curve inside [d_low, d_high] ---
                sig_win_x = sig_all_x[m_sig_all]
                sig_win_w = sig_all_w[m_sig_all]
                bkg_win_x = bkg_all_x[m_bkg_all]
                bkg_win_w = bkg_all_w[m_bkg_all]

                if len(sig_win_x) > 0 and len(bkg_win_x) > 0 and np.sum(sig_win_w) > 0 and np.sum(bkg_win_w) > 0:
                    y_true = np.concatenate([np.ones_like(sig_win_x), np.zeros_like(bkg_win_x)])
                    # Use -abs(dmet) as score so events closer to dmet=0 score higher (Signal-like)
                    y_scores = -np.abs(np.concatenate([sig_win_x, bkg_win_x]))
                    y_weights = np.concatenate([sig_win_w, bkg_win_w])

                    fpr, tpr, _ = roc_curve(y_true, y_scores, sample_weight=y_weights)
                    sorted_idx = np.argsort(fpr)
                    roc_auc = auc(fpr[sorted_idx], tpr[sorted_idx])

                    plt.figure(figsize=(10, 10))
                    plt.plot(fpr[sorted_idx], tpr[sorted_idx], lw=2, color='red', label=f'ROC curve (AUC = {roc_auc:.5f})')
                    plt.plot([0, 1], [0, 1], linestyle='--', color='gray', label='Random chance')
                    plt.xlabel("False Positive Rate")
                    plt.ylabel("True Positive Rate")
                    plt.title(f"ROC ({folder_tag}) | PV: {pct}% | [{d_low/1000:.0f}, {d_high/1000:.0f}] GeV")
                    plt.legend(loc="lower right")
                    plt.grid(True, which="both", linestyle="--", linewidth=0.5)
                    plt.tight_layout()

                    path2 = f"{base_out_dir}/roc_{file_tag}.png"
                    plt.savefig(path2, dpi=150)
                    plt.close()

        print(f"[{folder_tag}] Saved plots for Good PV scale = {pct}%")
```

```python
def getVarDict(fb, process, var_name=None):
    var_dict = {}

    if var_name is None or var_name == 'dmet':
        var_dict['dmet'] = {
            'var': fb['dmet'],
            'bins': np.linspace(-100000, 100000, 20+1),
            # 'bins': np.linspace(-250000, 250000, 40+1),
            'title': r'$E_{T,\mathrm{noJVT}}^{miss}-E_T^{miss}\ [MeV]$',
            'shift': '*1'
        }
    return var_dict

tot2 = tot
cut_name = 'basic'
var_config = getVarDict(tot2[0], 'ggHyyd', var_name="dmet")

# path for plot storage
mt_val_dir = '.'

for var in var_config:
    print("var: ", var)
    bg_values = []     
    bg_weights = []    
    bg_colors = []     
    bg_labels = [] 

    signal_values = [] 
    signal_weights = []
    signal_color = None 
    signal_label = None

    for j in range(len(ntuple_names)):
        process = ntuple_names[j]
        fb = tot2[j]  # TTree
        var_config = getVarDict(fb, process, var_name=var)

        x = var_config[var]['var'] # TBranch
        bins = var_config[var]['bins'] 
        weights = fb['weights']
        
        sample_info = sample_dict[process]
        color = sample_info['color']
        legend = sample_info['legend']

        
        if process == 'ggHyyd':  # signal
            signal_values.append(x)
            signal_weights.append(weights)
            signal_color = color
            signal_label = legend
        else:   # background
            bg_values.append(x)
            bg_weights.append(weights)
            bg_colors.append(color)
            bg_labels.append(legend)

    fig, (ax_top, ax_bot) = plt.subplots(2, 1, figsize=(12, 13), gridspec_kw={'height_ratios': [9, 4]})

    ax_top.hist(bg_values, bins=bins, weights=bg_weights, color=bg_colors,
                label=bg_labels, stacked=True)

    ax_top.hist(signal_values, bins=bins, weights=signal_weights, color=signal_color,
                label=signal_label, histtype='step', linewidth=2)

    signal_all = np.concatenate(signal_values) if len(signal_values) > 0 else np.array([])
    signal_weights_all = np.concatenate(signal_weights) if len(signal_weights) > 0 else np.array([])

    # Add error bar for signal (top plot)
    if len(signal_all) > 0:
        signal_counts, bin_edges = np.histogram(signal_all, bins=bins, weights=signal_weights_all)
        sum_weights_sq, _ = np.histogram(signal_all, bins=bins, weights=signal_weights_all**2)
        bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2
        signal_errors = np.sqrt(sum_weights_sq)  # Poisson error sqrt(N)

        ax_top.errorbar(bin_centers, signal_counts, yerr=signal_errors, fmt='.', linewidth=2,
                        color=signal_color, capsize=0)

    ax_top.set_yscale('log')
    ax_top.set_ylim(0.0001, 1e11)
    ax_top.set_xlim(bins[0], bins[-1])
    ax_top.minorticks_on()
    ax_top.grid(True, which="both", linestyle="--", linewidth=0.5)
    ax_top.set_ylabel("Events")
    ax_top.legend(ncol=2)

    bg_all = np.concatenate(bg_values) if len(bg_values) > 0 else np.array([])
    bg_weights_all = np.concatenate(bg_weights) if len(bg_weights) > 0 else np.array([])

    # Compute the weighted histogram counts using np.histogram
    S_counts, _ = np.histogram(signal_all, bins=bins, weights=signal_weights_all)
    B_counts, _ = np.histogram(bg_all, bins=bins, weights=bg_weights_all)     

    # Compute per-bin significance
    sig_simple = np.zeros_like(S_counts, dtype=float)
    sig_s_plus_b = np.zeros_like(S_counts, dtype=float)
    sig_s_plus_1p3b = np.zeros_like(S_counts, dtype=float)

    sqrt_B = np.sqrt(B_counts)
    sqrt_SplusB = np.sqrt(S_counts + B_counts)
    sqrt_Splus1p3B = np.sqrt(S_counts + 1.3 * B_counts)

    # Avoid division by zero safely
    sig_simple = np.where(B_counts > 0, S_counts / sqrt_B, 0)
    sig_s_plus_b = np.where((S_counts + B_counts) > 0, S_counts / sqrt_SplusB, 0)
    sig_s_plus_1p3b = np.where((S_counts + 1.3 * B_counts) > 0, S_counts / sqrt_Splus1p3B, 0)

    # Add Binomial ExpZ per bin
    zbi_per_bin = np.array([
        zbi(S_counts[i], B_counts[i], sigma_b_frac=0.3)
        for i in range(len(S_counts))
    ])

    # Compute the bin centers for plotting
    bin_centers = 0.5 * (bins[:-1] + bins[1:])

    # Compute the total significance: total S / sqrt(total B)
    total_signal = np.sum(S_counts)
    total_bkg = np.sum(B_counts)

    if total_bkg > 0:
        total_sig_simple = total_signal / np.sqrt(total_bkg)
        total_sig_s_plus_b = total_signal / np.sqrt(total_signal + total_bkg)
        total_sig_s_plus_1p3b = total_signal / np.sqrt(total_signal + 1.3 * total_bkg)
        total_sig_binomial = zbi(total_signal, total_bkg, sigma_b_frac=0.3)
    else:
        total_sig_simple = total_sig_s_plus_b = total_sig_s_plus_1p3b = total_sig_binomial = 0

    # --- Plot all significance curves ---
    ax_bot.step(bin_centers, sig_simple, where='mid', color='chocolate', linewidth=2,
                label=f"S/√B = {total_sig_simple:.4f}")
    ax_bot.step(bin_centers, sig_s_plus_b, where='mid', color='tomato', linewidth=2,
                label=f"S/√(S+B) = {total_sig_s_plus_b:.4f}")
    ax_bot.step(bin_centers, sig_s_plus_1p3b, where='mid', color='orange', linewidth=2,
                label=f"S/√(S+1.3B) = {total_sig_s_plus_1p3b:.4f}")
    ax_bot.step(bin_centers, zbi_per_bin, where='mid', color='plum', linewidth=2,
                label=f"Binomial ExpZ = {total_sig_binomial:.4f}")

    ax_bot.set_xlabel(var_config[var]['title'])
    # ax_bot.set_xticks(np.linspace(bins[0], bins[-1], 11))
    ax_bot.set_ylabel("Significance")
    # ax_bot.set_ylim(-0.8, 2)
    ax_top.set_xlim(bins[0], bins[-1])

    # Do not set a title on the bottom plot.
    ax_bot.set_title("")

    # Draw a legend with purple text.
    leg = ax_bot.legend()
    for text in leg.get_texts():
        text.set_color('purple')

    plt.xlim(bins[0], bins[-1])
    plt.tight_layout()
    path1 = f"/home/jlai/dark_photon/lumi300/plots/{var}.png"
    plt.savefig(path1)
    print(f"Successfully saved {var} performance plots to: ", path1)
    plt.close()
    plt.show()

    y_true = np.concatenate([np.ones_like(signal_all), np.zeros_like(bg_all)])
    y_scores = np.concatenate([signal_all, bg_all])
    # Combine the weights for all events.
    y_weights = np.concatenate([signal_weights_all, bg_weights_all])

    # Compute the weighted ROC curve.
    fpr, tpr, thresholds = roc_curve(y_true, y_scores, sample_weight=y_weights)
    sorted_indices = np.argsort(fpr)
    fpr_sorted = fpr[sorted_indices]
    tpr_sorted = tpr[sorted_indices]

    roc_auc = auc(fpr_sorted, tpr_sorted)

    # Create a new figure for the ROC curve.
    plt.figure(figsize=(8, 8))
    plt.plot(fpr, tpr, lw=2, color='red', label=f'ROC curve (AUC = {roc_auc:.5f})')
    plt.plot([0, 1], [0, 1], linestyle='--', color='gray', label='Random chance')
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title(f"ROC Curve for {var}")
    plt.legend(loc="lower right")
    plt.grid(True, which="both", linestyle="--", linewidth=0.5)
    plt.tight_layout()    
    path2 = f"/home/jlai/dark_photon/lumi300/plots/roc_curve_{var}.png"
    plt.savefig(path2)
    print(f"Successfully saved {var} roc curve plots to: ", path2)
    plt.close()
    plt.show()
```
